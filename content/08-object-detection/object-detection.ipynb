### **What is Object Detection?**

Object detection answers two questions at the same time:
- WHAT is in the image?
- WHERE is it?

Model response like this:

```
[
    {
        class: "person",
        confidence: 0.94,
        box: (120, 80, 300, 400)
    },

    {
        class: "car",
        confidence: 0.91,
        box: (400, 200, 700, 450)
    }
]
```

```
IMAGE
  ↓
Preprocessing
  ↓
MODEL
  ↓
Predictions
  ↓
┌─────────────────────────────┐
│ Person  0.94  [box]        │
│ Car     0.91  [box]        │
│ Dog     0.87  [box]        │
└─────────────────────────────┘

- YOLO predicts numerical values representing boxes, classes, and confidence scores.

| Concept              | Meaning                               |
| -------------------- | ------------------------------------- |
| Classification       | **What** is in the image?             |
| Localization         | **Where** is the main object?         |
| Detection            | **What + where** for multiple objects |
| Bounding box         | Rectangle around an object            |
| `(x1,y1,x2,y2)`      | Box represented by corners            |
| `(x,y,w,h)`          | Box represented by position + size    |
| Class                | Object category                       |
| Confidence           | How confident the model is            |
| Confidence threshold | Removes weak predictions              |


```
                 OBJECT DETECTION
                       │
             ┌─────────┴─────────┐
             ↓                   ↓
           WHAT?               WHERE?
             ↓                   ↓
          Class             Bounding Box
             │                   │
             └─────────┬─────────┘
                       ↓
                  Confidence

````python
import cv2
import numpy as np

# Create a blank image
image = np.zeros((500, 700, 3), dtype=np.uint8)

# Our "detector" says:
x1, y1 = 150, 100
x2, y2 = 450, 400

label = "Person"
confidence = 0.92

# Draw bounding box
cv2.rectangle(
    image,
    (x1, y1),
    (x2, y2),
    (0, 255, 0),
    2
)

# Draw label
text = f"{label} {confidence:.2f}"

cv2.putText(
    image,
    text,
    (x1, y1 - 10),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.7,
    (0, 255, 0),
    2
)

cv2.imshow("Object Detection", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

````python
def calculate_iou(box_a, box_b):

    # Intersection coordinates
    x1 = max(box_a[0], box_b[0])
    y1 = max(box_a[1], box_b[1])

    x2 = min(box_a[2], box_b[2])
    y2 = min(box_a[3], box_b[3])

    # Intersection dimensions
    intersection_width = max(0, x2 - x1)
    intersection_height = max(0, y2 - y1)

    intersection_area = (
        intersection_width * intersection_height
    )

    # Box A area
    area_a = (
        (box_a[2] - box_a[0]) *
        (box_a[3] - box_a[1])
    )

    # Box B area
    area_b = (
        (box_b[2] - box_b[0]) *
        (box_b[3] - box_b[1])
    )

    # Union
    union_area = area_a + area_b - intersection_area

    # IoU
    iou = intersection_area / union_area

    return iou
````

```
Predictions
     ↓
Sort by confidence
     ↓
Highest confidence → KEEP
     ↓
Calculate IoU with other boxes
     ↓
High overlap → SUPPRESS
     ↓
Next remaining box
     ↓
Final detections

### Next Part: Non-Maximum Suppression (NMS)

```
Predicted boxes
      ↓
Remove low-confidence boxes
      ↓
Sort by confidence
      ↓
Take highest-confidence box
      ↓
Keep it
      ↓
Calculate IoU with remaining boxes
      ↓
Remove boxes with high IoU
      ↓
Repeat
      ↓
Final boxes
```

- High confidence + high overlap = probably duplicate → suppress it.

````python
boxes = [
    [100, 100, 300, 400],  # A
    [110, 105, 295, 395],  # B
    [120, 120, 310, 410],  # C
    [500, 500, 700, 700]   # D
]

scores = [
    0.95,
    0.89,
    0.82,
    0.80
]
````

````python
iou_threshold = 0.5
````

````python
def nms(boxes, scores, iou_threshold=0.5):

    boxes = np.array(boxes)
    scores = np.array(scores)

    # Sort indices by confidence, highest first
    order = scores.argsort()[::-1]

    keep = []

    while len(order) > 0:

        # Highest-confidence box
        current = order[0]

        # Keep it
        keep.append(current)

        remaining = []

        for index in order[1:]:

            iou = calculate_iou(
                boxes[current],
                boxes[index]
            )

            if iou <= iou_threshold:
                remaining.append(index)

        order = np.array(remaining)

    return keep
````

````python
import numpy as np

boxes = [
    [100, 100, 300, 400],
    [110, 105, 295, 395],
    [120, 120, 310, 410],
    [500, 500, 700, 700]
]

scores = [
    0.95,
    0.89,
    0.82,
    0.80
]

keep = nms(
    boxes,
    scores,
    iou_threshold=0.5
)

print(keep)
````

- Box 1 → REMOVE
- Box 2 → REMOVE
- Box 3 → KEEP

### Full pipeline in video

````python
import cv2
import time
from ultralytics import YOLO
````

````python
model = YOLO("yolo11n.pt")
````

````python
video_path = "assets/people-walking.mp4"
output_path = "runs/people_detected.mp4"

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise RuntimeError("Could not open video")
````

````python
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"FPS: {fps}")
print(f"Resolution: {width}x{height}")
print(f"Frames: {frame_count}")
print(f"Duration: {frame_count / fps:.2f} sec")
````

````python
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

writer = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)
````

````python
frame_number = 0
start_time = time.time()

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # YOLO inference
    results = model(
        frame,
        classes=[0],      # person only
        conf=0.40,
        verbose=False
    )

    result = results[0]

    # Draw detections
    annotated_frame = result.plot()

    # FPS calculation
    elapsed = time.time() - start_time

    processing_fps = frame_number / elapsed

    cv2.putText(
        annotated_frame,
        f"FPS: {processing_fps:.1f}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    # Display
    cv2.imshow(
        "YOLO Person Detection",
        annotated_frame
    )

    # Save
    writer.write(annotated_frame)

    # ESC to stop
    if cv2.waitKey(1) & 0xFF == 27:
        break


cap.release()
writer.release()
cv2.destroyAllWindows()
````

```
Image
  ↓
Backbone
  ↓
Neck
  ↓
Detection Head
  ↓
Boxes + Classes + Confidence
```

- **Backbone**: image → edges → shapes → objects/features
- **Neck**: small objects, medium objects, and large objects
- **Head**: [x1, y1, x2, y2], confidence, class
