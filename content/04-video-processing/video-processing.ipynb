````python
import cv2

cap = cv2.VideoCapture("assets/people-walking.mp4")

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # 1. Resize
    frame = cv2.resize(frame, (640, 360))

    # 2. Grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 3. Blur
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # 4. Edge detection
    edges = cv2.Canny(blurred, 50, 150)

    cv2.imshow("Original", frame)
    cv2.imshow("Edges", edges)

    if cv2.waitKey(25) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
````

````python
import cv2

cap = cv2.VideoCapture("assets/people-walking.mp4")

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    "assets/processed_video.mp4",
    fourcc,
    fps,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Convert grayscale back to 3 channels
    gray_bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

    out.write(gray_bgr)

cap.release()
out.release()

print("Processed video saved!")
````

````python
import cv2

cap = cv2.VideoCapture("assets/people-walking.mp4")

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Process only every 2nd frame
    if frame_number % 2 == 0:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        cv2.imshow("Processed", gray)

        if cv2.waitKey(25) & 0xFF == ord("q"):
            break

    frame_number += 1

cap.release()
cv2.destroyAllWindows()
````

````python
import cv2
import time

cap = cv2.VideoCapture("assets/people-walking.mp4")

frame_number = 0
start_time = time.time()

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # Resize
    frame = cv2.resize(frame, (640, 360))

    # Process
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)

    # Calculate processing FPS
    elapsed = time.time() - start_time
    processing_fps = frame_number / elapsed if elapsed > 0 else 0

    # Add information to frame
    cv2.putText(
        edges,
        f"Frame: {frame_number}",
        (20, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        255,
        2
    )

    cv2.putText(
        edges,
        f"FPS: {processing_fps:.1f}",
        (20, 60),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        255,
        2
    )

    cv2.imshow("Real-Time Processing", edges)

    if cv2.waitKey(25) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
````

````python
import cv2

input_path = "assets/people-walking.mp4"
output_path = "assets/edge_video.mp4"

cap = cv2.VideoCapture(input_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = 640
height = 360

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame = cv2.resize(frame, (width, height))

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)

    # Convert edges to BGR because VideoWriter expects 3 channels
    output_frame = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

    frame_number += 1

    cv2.putText(
        output_frame,
        f"Frame: {frame_number}",
        (20, 35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 255),
        2
    )

    out.write(output_frame)

cap.release()
out.release()

print("Done!")
print(f"Saved: {output_path}")
````
